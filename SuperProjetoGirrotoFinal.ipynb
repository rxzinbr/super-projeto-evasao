```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('dataset.csv')

print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
print(df.columns.tolist())
```

```python
# Renomear colunas
df.rename(columns={
    'Gender': 'Gênero',
    'Target': 'Situacao'
}, inplace=True)

# Traduzir valores
df['Gênero'] = df['Gênero'].replace({
    1: 'Masculino',
    0: 'Feminino'
})

df['Scholarship holder'] = df['Scholarship holder'].replace({
    1: 'Sim',
    0: 'Não'
})

df['Situacao'] = df['Situacao'].replace({
    'Dropout': 'Evasão',
    'Graduate': 'Graduado',
    'Enrolled': 'Matriculado'
})
```

```python
df['Situacao'].value_counts()
```

```python
df['Média_Notas_Total'] = (
    df['Curricular units 1st sem (grade)']
    + df['Curricular units 2nd sem (grade)']
) / 2
```

```python
df['Faixa_Desempenho'] = pd.cut(
    df['Média_Notas_Total'],
    bins=[0, 10, 13, 16, 20],
    labels=['Baixo', 'Médio', 'Bom', 'Excelente']
)
```

```python
df['Faixa_Etaria'] = pd.cut(
    df['Age at enrollment'],
    bins=[15, 20, 25, 30, 40, 80],
    labels=['Até 20', '21 a 25', '26 a 30', '31 a 40', 'Acima de 40']
)
```

```python
#distribuição da situação academica
df['Situacao'].value_counts()
```

```python
#bolsa x situacao
(pd.crosstab(
    df['Scholarship holder'],
    df['Situacao'],
    normalize='index'
) * 100).round(2)
```

```python
#faixa desemp x situação
(pd.crosstab(
    df['Faixa_Desempenho'],
    df['Situacao'],
    normalize='index'
) * 100).round(2)
```

```python
#media das notas por situacao
df.groupby('Situacao')['Média_Notas_Total'].mean().round(2)
```

```python
#teorema bayes
p_evasao = (
    (df['Situacao'] == 'Evasão').mean()
)

print(p_evasao)
```

```python
#prob evasao desempenho baixo
p_evasao_baixo = (
    (
        (df['Faixa_Desempenho'] == 'Baixo') &
        (df['Situacao'] == 'Evasão')
    ).sum()
    /
    (df['Faixa_Desempenho'] == 'Baixo').sum()
)

print(p_evasao_baixo)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='Situacao'
)

plt.title('Distribuição da situação dos alunos')
plt.xlabel('Situação')
plt.ylabel('Quantidade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Situacao',
    y='Média_Notas_Total'
)

plt.title('Distribuição das notas por situação')
plt.xlabel('Situação')
plt.ylabel('Média das notas')

plt.show()
```

```python
tabela = pd.crosstab(
    df['Scholarship holder'],
    df['Situacao']
)

tabela.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Bolsa de estudos x situação')
plt.xlabel('Possui bolsa')
plt.ylabel('Quantidade')

plt.show()
```

```python
tabela = pd.crosstab(
    df['Faixa_Desempenho'],
    df['Situacao'],
    normalize='index'
) * 100

tabela.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Situação por faixa de desempenho')
plt.xlabel('Faixa de desempenho')
plt.ylabel('Percentual (%)')

plt.show()
```

```python
#faixa etaria x situacao
tabela = pd.crosstab(
    df['Faixa_Etaria'],
    df['Situacao'],
    normalize='index'
) * 100

tabela.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Situação por Faixa Etária')
plt.xlabel('Faixa Etária')
plt.ylabel('Percentual (%)')

plt.show()
```

```python
#distribuição geral das notas
plt.figure(figsize=(8,5))

sns.histplot(
    df['Média_Notas_Total'],
    bins=20,
    kde=True
)

plt.title('Distribuição das Médias dos Alunos')
plt.xlabel('Média das Notas')
plt.ylabel('Quantidade')

plt.show()
```
